###Ejercicio 1. Limpieza e imputación de un conjunto de datos mixto


```python
#importación de librerias

import numpy as np
import pandas as pd
```

```python
#creación del dataframe

df = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Pedro', None, 'Marta', 'Luis', 'Sofía'],
    'Edad': [25, np.nan, 35, 29, -5, 25, None],
    'Ciudad': ['Bogotá', 'Medellín', None, 'Medellín', 'Cali', 'Bogotá', 'Cali'],
    'Ingreso': [3500, 4800, np.nan, 5200, 5100, np.nan, 4700],
    'FechaIngreso': ['2023-01-01', '2023-01-05', None, '2023-01-10', '2023-01-12', None, '2023-01-15']
})
```

```python
df.info()

df
```

```python
#Detección de valores faltantes
df.isnull().sum()
```

```python
#Correción de valores faltantes columna "Nombre"
# Reemplazamos los valores faltantes (NaN) por la palabra "Desconocido"

df["Nombre"] = df["Nombre"].fillna("Desconocido")

#Correción de valores faltantes columna "Edad"
# Reemplazamos las edades negativas por NaN y luego llenamos los valores faltantes con la mediana.

df['Edad'] = df['Edad'].apply(lambda x: np.nan if x is None or x < 0 else x)
df['Edad'] = df['Edad'].fillna(df['Edad'].median())

#Correción de valores faltantes columna "Ciudad"
# Se reemplazan los valores faltantes con la moda (la ciudad que más se repite)

ciudad_mode = df['Ciudad'].mode(dropna=True)
if len(ciudad_mode) > 0:
    ciudad_mode = ciudad_mode[0]
else:
    ciudad_mode = 'Desconocida'
df['Ciudad'] = df['Ciudad'].fillna(ciudad_mode)

#Correción de valores faltantes columna "Ingreso" con la mediana
# Se usa la mediana porque es más robusta ante valores extremos que la media.

df['Ingreso'] = df['Ingreso'].fillna(df['Ingreso'].median())


#Correción de formato de fecha en columna "FechaIngreso"
# Convertimos las fechas a formato datetime y reemplazamos los valores nulos por la moda.

df['FechaIngreso'] = pd.to_datetime(df['FechaIngreso'], errors='coerce')
fecha_moda = df['FechaIngreso'].mode()[0]
df['FechaIngreso'] = df['FechaIngreso'].fillna(fecha_moda)

```

###Ejercicio 2. Limpieza de duplicados


```python
df = pd.DataFrame({
    'ID': [101, 102, 102, 103, 104, 104, 104],
    'Nombre': ['Ana', 'Luis', 'Luis', 'Marta', 'Carlos', 'Carlos', 'Carlos'],
    'Edad': [25, 30, 30, 29, 40, 40, 41],
    'Ciudad': ['Bogotá', 'Cali', 'Cali', 'Medellín', 'Cali', 'Cali', 'Cali'],
    'FechaRegistro': ['2023-01-01', '2023-01-05', '2023-01-05', '2023-01-10',
                    '2023-01-15', '2023-01-15', '2023-01-16']
})
```

```python
df
```

```python
total_registros = len(df)
print(f"Total de registros originales: {total_registros}")
```

```python
duplicados_exactos = df[df.duplicated(keep=False)]
print("Duplicados exactos:")
print(duplicados_exactos)

cantidad_duplicados_exactos = df.duplicated().sum()
print(f"Cantidad de duplicados exactos: {cantidad_duplicados_exactos}")
```

```python
duplicados_id_nombre = df[df.duplicated(subset=['ID', 'Nombre'], keep=False)]
print("Duplicados por ID y Nombre:")
print(duplicados_id_nombre)

cantidad_duplicados_id_nombre = df.duplicated(subset=['ID', 'Nombre']).sum()
print(f"Cantidad de duplicados por ID y Nombre: {cantidad_duplicados_id_nombre}")
```

```python
# Eliminamos duplicados conservando el primer registro
df_limpio = df.drop_duplicates(subset=['ID', 'Nombre'], keep='first')

registros_eliminados = len(df) - len(df_limpio)
print(f"Registros eliminados: {registros_eliminados}")
```

```python
print(f"Registros después de la limpieza: {len(df_limpio)}")
print("\nDataFrame limpio:")
df_limpio
```

###Ejercicio 3. Corrección de errores tipográficos o de codificación


```python

df = pd.DataFrame({
    'Ciudad': ['bogota', 'Bogotá', 'BOGOTA', 'bogotá', 'bogata', 'Bógota', 'BogoTa', 'Cali', 'calí', 'medellín', 'medellin']
})
```

```python
df
```

```python
# Pasamos todo a minúsculas y eliminamos espacios en blanco
df['Ciudad'] = df['Ciudad'].str.lower().str.strip()
```

```python
# Importamos función para eliminar acentos y aplicamos a cada valor

import unidecode
df['Ciudad'] = df['Ciudad'].apply(lambda x: unidecode.unidecode(x))
```

```python
# Definimos un diccionario con las correcciones
correcciones = {
    'bogata': 'bogota',}

# Reemplazamos según el diccionario

df['Ciudad'] = df['Ciudad'].replace(correcciones)
```
